```python
height=[3,0,2,0,4]
left=0
right=len(height)-1
left_max=0
right_max=0
water=0
while left<right:
    if height[left]< height[right]:
        if height[left]>=left_max:
            left_max=height[left]
        else:
            water+=left_max-height[left]
        left+=1
    else:
        if height[right]<height[left]:
            if height[right]>=right_max:
                right_max=height[right]
            else:
                water+=right_max-height[right]
            right-=1
print(water)
            
            
```

```python
sales = [2, 3, -2, 4]

curr_max = sales[0]
curr_min = sales[0]
max_product = sales[0]

for i in range(1, len(sales)):

    if sales[i] < 0:
        curr_max, curr_min = curr_min, curr_max

    curr_max = max(sales[i], curr_max * sales[i])
    curr_min = min(sales[i], curr_min * sales[i])

    max_product = max(max_product, curr_max)

print(max_product)
```

```python
products = input().split()

left = 0
max_length = 0
seen = set()

for right in range(len(products)):
    while products[right] in seen:
        seen.remove(products[left])
        left += 1

    seen.add(products[right])

    length = right - left + 1

    if length > max_length:
        max_length = length

print(max_length)
```

```python
transactions = list(map(int, input().split()))
target = int(input())

count = 0

for i in range(len(transactions)):
    total = 0

    for j in range(i, len(transactions)):
        total += transactions[j]

        if total == target:
            count += 1

print(count)
```

```python
skills = input().split()

groups = {}

for skill in skills:
    key = ''.join(sorted(skill))

    if key not in groups:
        groups[key] = []

    groups[key].append(skill)

for group in groups.values():
    print(group)
```

```python
packets = list(map(int, input().split()))

packet_set = set(packets)

max_length = 0

for num in packet_set:

    # Start only if num is the beginning
    # of a consecutive sequence
    if num - 1 not in packet_set:

        current = num
        length = 1

        while current + 1 in packet_set:
            current += 1
            length += 1

        if length > max_length:
            max_length = length

print(max_length)
```

```python
n = int(input())

appointments = []

for i in range(n):
    start, end = map(int, input().split())
    appointments.append([start, end])

appointments.sort()

merged = []

for appointment in appointments:

    start, end = appointment

    if not merged or start > merged[-1][1]:
        merged.append([start, end])

    else:
        merged[-1][1] = max(merged[-1][1], end)
print("Output:")

for start, end in merged:
    print(start, end)
```
